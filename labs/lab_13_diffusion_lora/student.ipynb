## Теоретическая часть

Latent diffusion добавляет шум в latent space VAE, а U-Net предсказывает noise/velocity. LoRA адаптирует attention projections без изменения всех весов. Фиксированные seeds и prompts обязательны для сравнения checkpoints.

![Архитектура latent diffusion](https://raw.githubusercontent.com/CompVis/latent-diffusion/main/assets/modelfigure.png)


Latent diffusion обучает UNet предсказывать ε в пространстве VAE. LoRA добавляет обучаемые низкоранговые поправки к attention projections; VAE, text encoder и base UNet остаются замороженными. CLIP text similarity, разнообразие и ближайшая train генерация измеряют разные свойства и требуют ручного дополнения.

Для LoRA $W^{\prime}=W+BA$, где $B\in\mathbb{R}^{d\times r}$, $A\in\mathbb{R}^{r\times k}$, $r\ll\min(d,k)$. Noise-prediction минимизирует $\mathbb{E}_{x_0,\epsilon,t}\|\epsilon-\epsilon_\theta(z_t,t,c)\|_2^2$ на VAE-latents.

### Baseline SD на фиксированных запросах

*Вес: 1 балл.*

**Постановка.** Baseline до обучения и сопоставимый inference protocol позволяют увидеть фактический вклад LoRA. Отдельная проверка этого шага не позволяет скрыть ошибку за итоговым показателем.

Выполните следующие действия:

1. Загрузите исходную SD1.5 на GPU и сгенерируйте baseline до установки адаптера на фиксированных запросах/seeds.

**Результат.** Проверяемое состояние «Baseline SD на фиксированных запросах» и названные в действии файлы при их создании.

**Критерий принятия.** Задание считается принятым, если выполнено действие «Загрузите исходную SD1.5 на GPU и сгенерируйте baseline до установки адаптера на фиксированных запросах/seeds» и его результат доступен следующему этапу.

```python
from diffusers import StableDiffusionPipeline, DPMSolverMultistepScheduler
import diffusers, transformers, peft
SD_ID = "stable-diffusion-v1-5/stable-diffusion-v1-5"
RES = 256 if FAST_DEV_RUN else 512
STEPS = 4 if FAST_DEV_RUN else 25
GUIDANCE = 7.5
pipe = StableDiffusionPipeline.from_pretrained(SD_ID, variant="fp16", torch_dtype=torch.float16, safety_checker=None, requires_safety_checker=False)
pipe.scheduler = DPMSolverMultistepScheduler.from_config(pipe.scheduler.config)
pipe = pipe.to(DEVICE)
pipe.set_progress_bar_config(disable=True)
GEN_DIR = ARTIFACTS / "generations"

def sync():
    if DEVICE == "mps": torch.mps.synchronize()
    elif DEVICE == "cuda": torch.cuda.synchronize()

@torch.no_grad()
def generate_set(tag):
    out_dir = GEN_DIR / tag; out_dir.mkdir(parents=True, exist_ok=True)
    rows, t0 = [], time.time()
    for pi, prompt in enumerate(EVAL_PROMPTS):
        for s in EVAL_SEEDS:
            path = out_dir / f"p{pi}_s{s}.png"
            if not path.exists():
                g = torch.Generator("cpu").manual_seed(s)       # одинаковый начальный шум для всех checkpoint
                img = pipe(prompt, num_inference_steps=STEPS, guidance_scale=GUIDANCE, height=RES, width=RES, generator=g).images[0]
                img.save(path)
            rows.append({"tag": tag, "prompt_id": pi, "prompt": prompt, "seed": s, "path": str(path)})
    sync()
    return rows, time.time() - t0

assert not getattr(pipe.unet, "peft_config", None), "baseline должен быть без LoRA"
baseline_rows, baseline_time = generate_set("baseline")
save_json("baseline.json", {"model": SD_ID, "scheduler": "DPMSolverMultistep", "steps": STEPS, "guidance": GUIDANCE, "resolution": RES,
                            "seconds": baseline_time, "images": baseline_rows})
print(f"baseline SD1.5 без адаптера: {len(baseline_rows)} изображений за {baseline_time:.0f} c ({baseline_time / len(baseline_rows):.1f} c/изобр.)")
print({"diffusers": diffusers.__version__, "transformers": transformers.__version__, "peft": peft.__version__, "torch": torch.__version__})
```

```python
from torch.utils.data import Dataset
import torchvision.transforms as T

class ButterflyDataset(Dataset):
    """Пара (изображение 512×512 в [-1,1], токены подписи CLIP, длина 77)."""
    def __init__(self, records, root, tokenizer, size):
        self.records, self.root, self.tok = records, Path(root), tokenizer
        self.tf = T.Compose([T.Resize(size, interpolation=T.InterpolationMode.BILINEAR), T.CenterCrop(size), T.ToTensor(), T.Normalize([0.5], [0.5])])
    def __len__(self):
        return len(self.records)
    def __getitem__(self, i):
        r = self.records[i]
        x = self.tf(Image.open(self.root / r["file_name"]).convert("RGB"))
        ids = self.tok(r["text"], padding="max_length", max_length=self.tok.model_max_length, truncation=True, return_tensors="pt").input_ids[0]
        return x, ids

train_ds = ButterflyDataset(meta, LORA_DATA, pipe.tokenizer, RES)
x0, ids0 = train_ds[0]
assert x0.shape == (3, RES, RES) and -1.0 <= float(x0.min()) and float(x0.max()) <= 1.0
assert ids0.shape == (77,) and int(ids0[0]) == pipe.tokenizer.bos_token_id
lens = [int((train_ds[i][1] != pipe.tokenizer.pad_token_id).sum()) for i in range(len(train_ds))]
print(f"датасет: {len(train_ds)} пар; изображение {tuple(x0.shape)} в [{float(x0.min()):.2f}, {float(x0.max()):.2f}]; длина подписи в токенах {min(lens)}–{max(lens)} (обрезка 77)")
```

### Оптимизация адаптеров и checkpoint

*Вес: 1 балл.*

**Постановка.** Различия rank и learning rate оцениваются при одинаковых изображениях, количестве optimizer updates и eval prompts. Отдельная проверка этого шага не позволяет скрыть ошибку за итоговым показателем.

Выполните следующие действия:

1. Обучите независимые адаптеры, сохраняйте optimizer, scaler и переносимые веса при каждом контрольном шаге.

**Результат.** Проверяемое состояние «Оптимизация адаптеров и checkpoint» и названные в действии файлы при их создании.

**Критерий принятия.** Задание считается принятым, если выполнено действие «Обучите независимые адаптеры, сохраняйте optimizer, scaler и переносимые веса при каждом контрольном шаге» и его результат доступен следующему этапу.

```python
from diffusers import DDPMScheduler
from peft.utils import get_peft_model_state_dict
from diffusers.utils import convert_state_dict_to_diffusers
import torch.nn.functional as F
noise_sched = DDPMScheduler.from_pretrained(SD_ID, subfolder="scheduler")

@torch.no_grad()
def cache_latents_and_text():
    lat, lat_flip, emb = [], [], []
    for x, ids in torch.utils.data.DataLoader(train_ds, batch_size=4):
        x = x.to(DEVICE, torch.float16)
        lat.append(pipe.vae.encode(x).latent_dist.mean * pipe.vae.config.scaling_factor)
        lat_flip.append(pipe.vae.encode(x.flip(-1)).latent_dist.mean * pipe.vae.config.scaling_factor)   # отражение — аугментация
        emb.append(pipe.text_encoder(ids.to(DEVICE))[0])
    return torch.cat(lat).float(), torch.cat(lat_flip).float(), torch.cat(emb)

LAT, LAT_FLIP, EMB = cache_latents_and_text()       # VAE и text encoder заморожены — латенты считаются один раз

def train_adapter(name, rank, lr):
    out = ARTIFACTS / "adapters" / name; out.mkdir(parents=True, exist_ok=True)
    unet = pipe.unet
    unet.add_adapter(LoraConfig(r=rank, lora_alpha=rank, init_lora_weights="gaussian", target_modules=TARGETS), adapter_name=name)
    unet.set_adapter(name)
    lora_params = [p for n, p in unet.named_parameters() if name in n and "lora" in n]
    for p in lora_params:
        p.data = p.data.float(); p.requires_grad_(True)   # обучаемые веса в fp32
    unet.enable_gradient_checkpointing(); unet.train()
    opt = torch.optim.AdamW(lora_params, lr=lr, weight_decay=1e-2)
    scaler = torch.amp.GradScaler(DEVICE)
    g = torch.Generator().manual_seed(SEED)               # одинаковый порядок данных и шум у всех конфигураций
    hist, step, t0 = [], 0, time.time()
    resume = sorted(out.glob("step_*.pt"), key=lambda p: int(p.stem.split("_")[1]))
    if resume:
        ck = torch.load(resume[-1], map_location=DEVICE, weights_only=False)
        for p, v in zip(lora_params, ck["lora"]): p.data.copy_(v)
        opt.load_state_dict(ck["optimizer"]); scaler.load_state_dict(ck["scaler"]); hist, step = ck["loss"], ck["step"]; g.set_state(ck["rng"].cpu())
        print(f"{name}: продолжаю с шага {step}")
    while step < UPDATES:
        idx = torch.randint(0, len(LAT), (BATCH,), generator=g)
        flip = torch.rand(BATCH, generator=g) < 0.5
        z0 = torch.where(flip[:, None, None, None].to(DEVICE), LAT_FLIP[idx.to(DEVICE)], LAT[idx.to(DEVICE)])
        noise = torch.randn(z0.shape, generator=g).to(DEVICE)
        t = torch.randint(0, noise_sched.config.num_train_timesteps, (BATCH,), generator=g).to(DEVICE)
        zt = noise_sched.add_noise(z0, noise, t)
        for pg in opt.param_groups:
            pg["lr"] = lr * min(1.0, (step + 1) / 20)
        with torch.autocast(DEVICE, dtype=torch.float16):
            pred = unet(zt.half(), t, EMB[idx.to(DEVICE)]).sample
        loss = F.mse_loss(pred.float(), noise)
        opt.zero_grad(set_to_none=True)
        scaler.scale(loss).backward()
        scaler.unscale_(opt); torch.nn.utils.clip_grad_norm_(lora_params, 1.0)
        scaler.step(opt); scaler.update()
        hist.append(loss.item()); step += 1
        if step in CKPT_AT:
            sd = convert_state_dict_to_diffusers(get_peft_model_state_dict(unet, adapter_name=name))
            StableDiffusionPipeline.save_lora_weights(out / f"step_{step}", unet_lora_layers=sd, safe_serialization=True)
            torch.save({"lora": [p.detach().clone() for p in lora_params], "optimizer": opt.state_dict(), "scaler": scaler.state_dict(),
                        "loss": hist, "step": step, "rng": g.get_state()}, out / f"step_{step}.pt")
            print(f"{name}: шаг {step}, loss (скользящее среднее 50) {np.mean(hist[-50:]):.4f}, {time.time() - t0:.0f} c")
    unet.eval(); unet.disable_gradient_checkpointing()
    unet.delete_adapters(name)                       # следующий адаптер учится с нуля от той же основы
    return {"loss": hist, "seconds": time.time() - t0, "updates": step}

train_stats = {}
for name, cfg in CONFIGS.items():
    torch.manual_seed(SEED)
    train_stats[name] = train_adapter(name, cfg["rank"], cfg["lr"])
save_json("train_stats.json", {k: {"updates": v["updates"], "seconds": v["seconds"], "loss": v["loss"]} for k, v in train_stats.items()})
import pandas as pd
print(pd.DataFrame({k: {"updates": v["updates"], "минут": v["seconds"] / 60, "loss первые 50": np.mean(v["loss"][:50]), "loss последние 50": np.mean(v["loss"][-50:])}
                    for k, v in train_stats.items()}).T.round(4).to_string())
```

### Признаки CLIP для независимой оценки

*Вес: 1 балл.*

**Постановка.** Одна similarity-метрика не определяет качество генерации: высокое совпадение с текстом может сопровождаться копированием train. Отдельная проверка этого шага не позволяет скрыть ошибку за итоговым показателем.
